## Saving results into cloud

`````python
print(EXPORT_FOLDER)
`````

`````python
import shutil
import os

# 2. Configurar las rutas completas
# Origen (usualmente en el almacenamiento local de la instancia)
source_path = f"{ROOT_PATH}/outputs"

# Destino (dentro de tu Google Drive montado)
destination_path = f"/content/drive/MyDrive/{EXPORT_FOLDER}"

# 3. Ejecutar el movimiento
try:
    if not os.path.exists(destination_path):
        os.makedirs(destination_path)
        print(f"Carpeta creada: {destination_path}")
    # shutil.move gestiona la copia y eliminación del original
    # incluso entre diferentes sistemas de archivos (de local a Drive)
    shutil.move(source_path, destination_path)
    print(f"Se ha movido correctamente a: {destination_path}")
except FileNotFoundError:
    print(f"Error: No se encontró la carpeta de origen en {source_path}")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
`````
